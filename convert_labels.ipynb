```python
import numpy as np
from pathlib import Path
import json
```

```python
label_path = Path("/home/johnz/projects/tools/SynthSeg/data/labels_classes_priors")
out_path = Path("/home/johnz/projects/tools/SynthSegFast/synthseg_fast/res/constants/")
v2 = {}
v1 = {}
for label in label_path.glob("synthseg*.npy"):
    label_name = label.stem
    label = np.load(label)
    o = v2 if "2.0" in label_name else v1
    label_name = label_name.replace("synthseg_", "").replace("_2.0", "")
    category, t = label_name.split("_")
    if category not in o:
        o[category] = {}
    o[category][t] = label.tolist()
    
```

```python
for folder, data in [(out_path/"v1.0", v1), (out_path/"v2.0", v2)]:
    folder.mkdir(exist_ok=True, parents=True)
    for category in data:
        with (folder/Path(category+".json")).open("w") as f:
            json.dump(data[category], f, indent=4)

```
